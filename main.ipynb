```python
# import all pakcgaes 
import numpy as np 
import matplotlib.pyplot as plt 
import random 
%matplotlib inline 

# sklearn modules 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
# import dataset from ubyte files 

def read_mnist_images(file_path):
    with open(file_path, 'rb') as file : 
        magic, num_images, rows, cols = np.fromfile(file, dtype=">i4" , count=4)
        images = np.fromfile(file,dtype=np.uint8)
        images = images.reshape(num_images, rows, cols)
        return images


def read_mnist_labels(file_path):
    with open(file_path, 'rb') as file:
        magic, num_items = np.fromfile(file, dtype=">i4", count=2)
        labels = np.fromfile(file, dtype=np.uint8)
        return labels

# make function to show images randomly 
def show_random_image_from_array(images_array):
    num_images = images_array.shape[0]

    random_index = random.randint(0, num_images - 1) 
    random_image = images_array[random_index]

    plt.figure(figsize=(4,4))
    plt.imshow(random_image)
    plt.title(f"random image index: {random_index}")
    plt.show()
```

```python
# read the data 

TRAIN_IMAGES_PATH = "dataset/train-images-idx3-ubyte/train-images-idx3-ubyte"
TRAIN_LABELES_PATH = "dataset/train-labels-idx1-ubyte/train-labels-idx1-ubyte"

TEST_IMAGES_PATH = "dataset/t10k-images-idx3-ubyte/t10k-images-idx3-ubyte"
TEST_LABELES_PATH = "dataset/t10k-labels-idx1-ubyte/t10k-labels-idx1-ubyte"

# read train data
train_images = read_mnist_images(TRAIN_IMAGES_PATH)
train_labels = read_mnist_labels(TRAIN_LABELES_PATH)

# read test data
test_images = read_mnist_images(TEST_IMAGES_PATH)
test_labels = read_mnist_labels(TEST_LABELES_PATH)
```
